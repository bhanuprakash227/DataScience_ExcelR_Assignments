#**Hospital Patient Data Analysis**

**Patient Dataset**

```python
import pandas as pd
import numpy as np

patient_df = pd.read_csv("Patient_Data.csv")

print("Patient Dataset:")
print(patient_df)

print("\nDataset Information:")
patient_df.info()
```

**Billing-Related Columns**

```python
billing_columns = [
    "PatientID",
    "Department",
    "Doctor",
    "BillAmount"
]

billing_patient_df = patient_df[billing_columns].copy()

print("Billing Related Data:")
print(billing_patient_df)
```

**Drop Administrative Columns**

```python
clean_patient_df = patient_df.drop(
    columns=["ReceptionistID", "CheckInTime"]
)

print("Dataset after dropping administrative columns:")
print(clean_patient_df)
```

**total bill amount per department**

```python
department_bill = (
    patient_df.groupby("Department")["BillAmount"]
    .sum()
    .reset_index()
)

department_bill.columns = ["Department", "TotalBillAmount"]

print("Total Bill Amount by Department:")
print(department_bill)
```

**Remove duplicate**

```python
unique_patients = patient_df.drop_duplicates(
    subset="PatientID"
)

print("Patient Records after Removing Duplicates:")
print(unique_patients)
```

**mean bill**

```python
mean_bill = patient_df["BillAmount"].mean()
print("Mean Bill Amount:", mean_bill)

# Fill missing bill amounts with mean

patient_df["BillAmount"] = patient_df["BillAmount"].fillna(mean_bill)
print("\nDataset after filling missing BillAmount:")
print(patient_df)
```

**Load billing dataset**

```python
billing_df = pd.read_csv("Billing_Data.csv")
print("Billing Dataset:")
print(billing_df)

# Merge patient and billing datasets
merged_df = pd.merge(
    patient_df,
    billing_df,
    on="PatientID",
    how="left"
)

print("\nMerged Dataset:")
print(merged_df)
```

**Create new patients**

```python
new_patients = pd.DataFrame({
    "PatientID": [107, 108],
    "Name": ["Frank", "Grace"],
    "Department": ["Cardiology", "Neurology"],
    "Doctor": ["Dr. Smith", "Dr. John"],
    "BillAmount": [6500, 7200],
    "ReceptionistID": [4, 5],
    "CheckInTime": ["2023-01-16 09:30", "2023-01-17 10:00"]
})
print("New Patients:")
print(new_patients)

# Concatenate row-wise
patients_with_new = pd.concat(
    [patient_df, new_patients],
    ignore_index=True
)
print("\nDataset after adding new patients:")
print(patients_with_new)
```

**new billing**

```python
new_billing_columns = pd.DataFrame({
    "InsuranceCovered": [2000, 1500, 2500, 3000, 1000, 1800],
    "FinalAmount": [3000, 3500, 5000, 3200, 4000, 4700]
})

patient_data_for_columns = patient_df.reset_index(drop=True)
new_billing_columns = new_billing_columns.reset_index(drop=True)

final_dataset = pd.concat(
    [patient_data_for_columns, new_billing_columns],
    axis=1
)
print("Final Dataset:")
print(final_dataset)
```

**FINAL CLEANED HOSPITAL DATASET**

```python
final_cleaned = patient_df.copy()

final_cleaned = final_cleaned.drop(
    columns=["ReceptionistID", "CheckInTime"]
)

mean_bill = final_cleaned["BillAmount"].mean()

final_cleaned["BillAmount"] = (
    final_cleaned["BillAmount"]
    .fillna(mean_bill)
)

final_cleaned = final_cleaned.drop_duplicates(
    subset="PatientID"
)

final_cleaned = pd.merge(
    final_cleaned,
    billing_df,
    on="PatientID",
    how="left"
)

print("FINAL CLEANED DATASET")
print("=" * 50)
print(final_cleaned)
```

**Department-wise revenue**

```python
department_revenue = (
    final_cleaned
    .groupby("Department")["BillAmount"]
    .sum()
    .sort_values(ascending=False)
)

print("Department-Wise Revenue:")
print(department_revenue)
```

**Doctor-wise billing performance**

```python
doctor_performance = (
    final_cleaned
    .groupby("Doctor")["BillAmount"]
    .agg(["count", "sum", "mean"])
    .reset_index()
)

doctor_performance.columns = [
    "Doctor",
    "Number_of_Patients",
    "Total_Bill",
    "Average_Bill"
]

print("Doctor-Wise Performance:")
print(doctor_performance)
```
