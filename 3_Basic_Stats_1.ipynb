**1. Import Libraries**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

**2. Load the Dataset**

```python
df = pd.read_csv("sales_data_with_discounts.csv")

print("Dataset loaded successfully!")

print("\nFirst 5 rows:")
print(df.head())
```

**3. Understand the Dataset**

```python
print("Shape of Dataset:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

**4. Identify Numerical and Categorical Columns**

```python
#numerical columns
numerical_columns = df.select_dtypes(include=np.number).columns
print("Numerical Columns:")
print(list(numerical_columns))

#categorical columns
categorical_columns = df.select_dtypes(include="object").columns
print("\nCategorical Columns:")
print(list(categorical_columns))
```

**5. Descriptive Statistics**

```python
print("----- MEAN -----")
print(df[numerical_columns].mean())

print("\n----- MEDIAN -----")
print(df[numerical_columns].median())

print("\n----- MODE -----")
print(df[numerical_columns].mode().iloc[0])

print("\n----- STANDARD DEVIATION -----")
print(df[numerical_columns].std())
```

**6. Complete Statistical Summary**

```python
print(df[numerical_columns].describe())
```

**7. Interpretation of Statistics**

```python
print("INTERPRETATION")
print("----------------")

for column in numerical_columns:
    mean_value = df[column].mean()
    median_value = df[column].median()
    std_value = df[column].std()

    print(f"\n{column}:")
    print(f"Mean = {mean_value:.2f}")
    print(f"Median = {median_value:.2f}")
    print(f"Standard Deviation = {std_value:.2f}")

    if mean_value > median_value:
        print("The mean is greater than the median, indicating possible right skewness.")
    elif mean_value < median_value:
        print("The mean is less than the median, indicating possible left skewness.")
    else:
        print("The mean and median are approximately equal.")
```

**8. Histograms**

```python
for column in numerical_columns:
    plt.figure(figsize=(8, 5))

    plt.hist(df[column], bins=20, edgecolor="black")

    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Frequency")

    plt.show()
```

**9. Histogram Interpretation**

```python
print("HISTOGRAM ANALYSIS")
print("------------------")

for column in numerical_columns:
    skewness = df[column].skew()

    print(f"\n{column}")
    print(f"Skewness: {skewness:.2f}")

    if skewness > 0.5:
        print("Inference: The distribution is positively skewed.")
    elif skewness < -0.5:
        print("Inference: The distribution is negatively skewed.")
    else:
        print("Inference: The distribution is approximately symmetric.")
```

**10. Box Plots**

```python
for column in numerical_columns:
    plt.figure(figsize=(8, 5))

    plt.boxplot(df[column], vert=True)

    plt.title(f"Box Plot of {column}")
    plt.ylabel(column)

    plt.show()
```

**11. Detect Outliers Using IQR**

```python
print("OUTLIER ANALYSIS")
print("----------------")

for column in numerical_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    print(f"\nColumn: {column}")
    print(f"Q1: {Q1:.2f}")
    print(f"Q3: {Q3:.2f}")
    print(f"IQR: {IQR:.2f}")
    print(f"Lower Limit: {lower_limit:.2f}")
    print(f"Upper Limit: {upper_limit:.2f}")
    print(f"Number of Outliers: {len(outliers)}")
```

**12. Categorical Column Analysis**

```python
print("Categorical Columns:")
print(list(categorical_columns))
```

**13. Bar Charts for Categorical Columns**

```python
for column in categorical_columns:

    plt.figure(figsize=(10, 5))

    value_counts = df[column].value_counts()

    value_counts.plot(kind="bar")

    plt.title(f"Category Distribution - {column}")
    plt.xlabel(column)
    plt.ylabel("Count")

    plt.xticks(rotation=45)
    plt.tight_layout()

    plt.show()
```

**14. Categorical Distribution Analysis**

```python
print("CATEGORICAL DATA ANALYSIS")
print("-------------------------")

for column in categorical_columns:

    print(f"\nColumn: {column}")

    counts = df[column].value_counts()

    print(counts)

    print("Most frequent category:", counts.index[0])
    print("Frequency:", counts.iloc[0])
```

**15. Missing Values Check**

```python
print("Missing Values:")
print(df.isnull().sum())
```

**16. Final Summary**

```python
print("=" * 60)
print("BASIC STATISTICS — FINAL SUMMARY")
print("=" * 60)

print("\nDataset Shape:")
print(df.shape)

print("\nNumerical Columns:")
print(list(numerical_columns))

print("\nCategorical Columns:")
print(list(categorical_columns))

print("\nMissing Values:")
print(df.isnull().sum().sum())

print("\nStatistical Summary:")
print(df[numerical_columns].describe())
```
