#**EXPLORATORY DATA ANALYSIS**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
print("Libraries imported successfully.")
```

```python
df = pd.read_csv("Cardiotocographic.csv")
print("Dataset loaded successfully.")
print("\nFirst 5 rows:")
display(df.head())
```

```python
print("Dataset Shape:")
print(df.shape)
print("\nColumn Names:")
print(df.columns.tolist())
print("\nDataset Information:")
df.info()
```

**Statistical overview**

```python
print("Descriptive Statistics:")
display(df.describe(include="all"))
```

**Check missing values**

```python
missing_values = df.isnull().sum()
print("Missing Values in Each Column:")
print(missing_values)
print("\nTotal Missing Values:")
print(missing_values.sum())
```

**Handle missing values**

```python
numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(df[column].median())

print("Missing values handled using median imputation.")

print("\nRemaining Missing Values:")
print(df.isnull().sum().sum())
```

**Check data types**

```python
print("Data Types:")
print(df.dtypes)
```

```python
for column in df.columns:
    converted = pd.to_numeric(df[column], errors="coerce")

    if converted.notna().mean() >= 0.90:
        df[column] = converted

print("Data types after conversion:")
print(df.dtypes)
```

**Identify numerical columns**

```python
numeric_columns = df.select_dtypes(include=np.number).columns

print("Numerical Columns:")
print(list(numeric_columns))
```

**Calculate mean , median , standard deviation**

```python
print("=" * 50)
print("MEAN")
print("=" * 50)

for column in numeric_columns:
    print(f"{column}: {df[column].mean():.2f}")

print("=" * 50)
print("MEDIAN")
print("=" * 50)

for column in numeric_columns:
    print(f"{column}: {df[column].median():.2f}")

print("=" * 50)
print("STANDARD DEVIATION")
print("=" * 50)

for column in numeric_columns:
    print(f"{column}: {df[column].std():.2f}")
```

**Calculate IQR**

```python
print("INTERQUARTILE RANGE (IQR)")
print("=" * 50)

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    print(f"{column}: {IQR:.2f}")
```

**complete statistical summary**

```python
summary = pd.DataFrame(index=numeric_columns)

summary["Mean"] = df[numeric_columns].mean()
summary["Median"] = df[numeric_columns].median()
summary["Standard Deviation"] = df[numeric_columns].std()
summary["Q1"] = df[numeric_columns].quantile(0.25)
summary["Q3"] = df[numeric_columns].quantile(0.75)
summary["IQR"] = summary["Q3"] - summary["Q1"]

print("COMPLETE STATISTICAL SUMMARY")
display(summary)
```

**outliers**

```python
print("OUTLIER ANALYSIS")
print("=" * 60)

outlier_summary = []

for column in numeric_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    outlier_summary.append({
        "Column": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Limit": lower_limit,
        "Upper Limit": upper_limit,
        "Outliers": len(outliers)
    })

outlier_df = pd.DataFrame(outlier_summary)

display(outlier_df)
```

**Histogram**

```python
for column in numeric_columns:

    plt.figure(figsize=(8, 5))

    plt.hist(df[column], bins=20, edgecolor="black")

    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Frequency")

    plt.tight_layout()
    plt.show()
```

**Boxplots**

```python
for column in numeric_columns:

    plt.figure(figsize=(8, 5))

    sns.boxplot(x=df[column])

    plt.title(f"Boxplot of {column}")
    plt.xlabel(column)

    plt.show()
```

**skewness**

```python
print("SKEWNESS ANALYSIS")
print("=" * 50)

for column in numeric_columns:

    skewness = df[column].skew()

    print(f"\n{column}: {skewness:.2f}")

    if skewness > 0.5:
        print("Distribution is positively skewed.")
    elif skewness < -0.5:
        print("Distribution is negatively skewed.")
    else:
        print("Distribution is approximately symmetric.")
```

**categorical columns**

```python
categorical_columns = df.select_dtypes(
    include=["object", "category"]
).columns

print("Categorical Columns:")
print(list(categorical_columns))
```

**Bar charts**

```python
for column in categorical_columns:

    plt.figure(figsize=(10, 5))

    df[column].value_counts().plot(kind="bar")

    plt.title(f"Category Frequency - {column}")
    plt.xlabel(column)
    plt.ylabel("Frequency")

    plt.xticks(rotation=45)
    plt.tight_layout()

    plt.show()
```

**correlation matrix**

```python
correlation_matrix = df[numeric_columns].corr()

print("Correlation Matrix:")
display(correlation_matrix)
```

**Correlation heatmap**

```python
plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap")

plt.tight_layout()
plt.show()
```

```python
print("STRONG CORRELATIONS")
print("=" * 50)

corr_pairs = correlation_matrix.unstack()

strong_correlations = corr_pairs[
    (abs(corr_pairs) >= 0.70) &
    (abs(corr_pairs) < 1.00)
].sort_values(ascending=False)

print(strong_correlations)
```

```python
corr_matrix = df[numeric_columns].corr()

corr_pairs = corr_matrix.abs().unstack()

corr_pairs = corr_pairs[
    corr_pairs < 1
]

if len(corr_pairs) > 0:

    strongest_pair = corr_pairs.idxmax()

    col1 = strongest_pair[0]
    col2 = strongest_pair[1]

    print("Strongest relationship:")
    print(col1, "and", col2)

    plt.figure(figsize=(8, 5))

    plt.scatter(
        df[col1],
        df[col2],
        alpha=0.6
    )

    plt.xlabel(col1)
    plt.ylabel(col2)
    plt.title(f"{col1} vs {col2}")

    plt.show()
```

**Pair plot**

```python
# Use a limited number of columns if the dataset has many variables
pair_columns = list(numeric_columns[:6])

sns.pairplot(
    df[pair_columns]
)

plt.show()
```

**Violin plots**

```python
for column in numeric_columns[:6]:

    plt.figure(figsize=(8, 5))

    sns.violinplot(x=df[column])

    plt.title(f"Violin Plot of {column}")
    plt.xlabel(column)

    plt.show()
```

```python
print("=" * 60)
print("PATTERN RECOGNITION AND INSIGHTS")
print("=" * 60)

for column in numeric_columns:

    mean = df[column].mean()
    median = df[column].median()
    std = df[column].std()

    print(f"\nColumn: {column}")
    print(f"Mean: {mean:.2f}")
    print(f"Median: {median:.2f}")
    print(f"Standard Deviation: {std:.2f}")

    if mean > median:
        print("Pattern: Mean is greater than median.")
        print("This may indicate positive/right skewness.")

    elif mean < median:
        print("Pattern: Mean is less than median.")
        print("This may indicate negative/left skewness.")

    else:
        print("Pattern: Mean and median are approximately equal.")
```

```python
print("Columns available in the dataset:")
for i, column in enumerate(df.columns, start=1):
    print(i, ".", column)
```

**Final EDA Report**

```python
print("=" * 70)
print("EXPLORATORY DATA ANALYSIS — FINAL REPORT")
print("=" * 70)

print("\n1. DATASET OVERVIEW")
print("-------------------")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\n2. NUMERICAL VARIABLES")
print("----------------------")
print(list(numeric_columns))

print("\n3. CATEGORICAL VARIABLES")
print("------------------------")
print(list(categorical_columns))

print("\n4. MISSING VALUES")
print("-----------------")
print("Total missing values:", df.isnull().sum().sum())

print("\n5. STATISTICAL SUMMARY")
print("----------------------")
display(summary)

print("\n6. CORRELATION INSIGHTS")
print("-----------------------")

corr_matrix = df[numeric_columns].corr()

for i in range(len(corr_matrix.columns)):
    for j in range(i + 1, len(corr_matrix.columns)):

        correlation = corr_matrix.iloc[i, j]

        if abs(correlation) >= 0.70:
            print(
                f"{corr_matrix.columns[i]} and "
                f"{corr_matrix.columns[j]}: "
                f"{correlation:.2f}"
            )
```
