#**Confidence Interval**

```python
import numpy as np
from scipy import stats

data=np.array([1.13,1.55,1.43,0.92,1.25,1.36,1.32,0.85,1.07,1.48,1.20,1.33,1.18,1.22,1.29])
n=len(data)

mean=np.mean(data)
sample_std=np.std(data,ddof=1)
alpha=0.01
t_critical=stats.t.ppf(1-alpha/2,n-1)
margin_t=t_critical*(sample_std/np.sqrt(n))
lower_t=mean-margin_t
upper_t=mean+margin_t

print("PART A: 99% Confidence Interval Using Sample Standard Deviation")
print("Sample Size:",n)
print("Sample Mean:",mean)
print("Sample Standard Deviation:",sample_std)
print("Degrees of Freedom:",n-1)
print("t-Critical Value:",t_critical)
print("Margin of Error:",margin_t)
print("99% Confidence Interval:",(lower_t,upper_t))
print("Rationale: The population standard deviation is unknown and the sample size is small (n=15), so the t-distribution is used.")

population_std=0.2
z_critical=stats.norm.ppf(1-alpha/2)
margin_z=z_critical*(population_std/np.sqrt(n))
lower_z=mean-margin_z
upper_z=mean+margin_z

print("\nPART B: 99% Confidence Interval Using Known Population Standard Deviation")
print("Known Population Standard Deviation:",population_std)
print("Sample Size:",n)
print("Sample Mean:",mean)
print("z-Critical Value:",z_critical)
print("Margin of Error:",margin_z)
print("99% Confidence Interval:",(lower_z,upper_z))
```

#**Hypothesis Testing**

```python
import math
from scipy import stats

n=25
sample_mean=3050
mu=1000+5*600
sigma=5*25
alpha=0.05

print("1. Hypotheses Statement")
print("H0: The mean weekly operating cost is equal to the theoretical cost of Rs.",mu)
print("H1: The mean weekly operating cost is greater than the theoretical cost of Rs.",mu)

z=(sample_mean-mu)/(sigma/math.sqrt(n))
print("\n2. Test Statistic")
print("Sample Mean:",sample_mean)
print("Theoretical Mean:",mu)
print("Standard Deviation:",sigma)
print("Sample Size:",n)
print("Z Test Statistic:",z)

p_value=1-stats.norm.cdf(z)
critical_value=stats.norm.ppf(1-alpha)
print("\n3. Probability and Comparison")
print("Alpha:",alpha)
print("P-Value:",p_value)
print("Critical Value:",critical_value)
print("\n4. Decision")

if z>critical_value:
    print("Since the test statistic is greater than the critical value, reject H0.")
    decision="Reject H0"
else:
    print("Since the test statistic is not greater than the critical value, fail to reject H0.")
    decision="Fail to Reject H0"


print("\n5. Conclusion")
if decision=="Reject H0":
    print("There is strong evidence to support the restaurant owners' claim that weekly operating costs are higher than the model suggests.")
else:
    print("There is not enough evidence to support the restaurant owners' claim that weekly operating costs are higher than the model suggests.")
```
