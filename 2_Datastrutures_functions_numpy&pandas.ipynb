#**Section A — Data Structures & Control Structures**

**1: List Operations**

```python
numbers = [50, 20, 40, 20, 10]
print("Original List:", numbers)

numbers.append(60)
print("After append:", numbers)

numbers.extend([70, 80])
print("After extend:", numbers)

numbers.insert(1, 25)
print("After insert:", numbers)

numbers.remove(20)
print("After remove:", numbers)

removed_element = numbers.pop()
print("Popped element:", removed_element)
print("After pop:", numbers)

print("Index of 40:", numbers.index(40))

print("Count of 20:", numbers.count(20))

numbers.sort()
print("After sort:", numbers)

numbers.reverse()
print("After reverse:", numbers)

numbers.clear()
print("After clear:", numbers)
```

**2:Tuple**

```python
students = ("Rahul", "Priya", "Arun")
print("Original Tuple:", students)

try:
    students[1] = "Bhanu"
except TypeError as e:
    print("Error:", e)

print("\nExplanation:")
print("Tuples are immutable, so their elements cannot be changed after creation.")
```

**3:Set**

```python
numbers = {10, 20, 30, 20, 40, 10, 50, 30}

print("Set:", numbers)

print("\nExplanation:A set automatically removes duplicate values.Therefore, each value appears only once.")
```

**4:Dictionary**

```python
student = {
    "name": "Bhanu",
    "age": 22,
    "city": "Warangal"
}
student["age"] = 23
student["email"] = "bhanu@gmail.com"

print("Final Dictionary:")
print(student)
```

**5:Voting Elgibility**

```python
age =int(input("Enter the age:"))
if age >= 18:
    print("The person is eligible to vote.")
else:
    print("The person is not eligible to vote.")
```

**6:Grade**

```python
marks =int(input("Enter the marks:"))
if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
elif marks >= 50:
    grade = "C"
else:
    grade = "Fail"
print("Marks:", marks)
print("Grade:", grade)
```

**7:Positive,Negitive or Zero**

```python
number = int(input("Enter a number: "))
if number > 0:
    print("The number is positive.")
    if number % 2 == 0:
        print("The number is even.")
    else:
        print("The number is odd.")
elif number == 0:
    print("The number is zero.")
else:
    print("The number is negative.")
```

#**Section B — NumPy**

**NumPy Arrays**

```python
import numpy as np
scalar = np.array(5)                      #scaler
array_1d = np.array([1, 2, 3, 4, 5])      #1D
array_2d = np.array([
    [10, 20, 30],
    [40, 50, 60]
])                                        #2d

print("Scalar:")
print(scalar)

print("\n1D Array:")
print(array_1d)

print("\n2D Array:")
print(array_2d)
```

**Random 4×4 NumPy Array**

#**Section C — Exploring Pandas**

**Pandas DataFrame Operations**

```python
import pandas as pd
import numpy as np
data = {
    "Name": [
        "Rahul", "Priya", "Arun", "Bhanu", "Sneha",
        "Kiran", "Anjali", "Vijay", "Rahul", "Priya"
    ],

    "Age": [
        22, 24, 23, 25, 21,
        26, 24, 27, 22, 24
    ],

    "City": [
        "Warangal", "Hyderabad", "Delhi", "Warangal", "Mumbai",
        "Chennai", "Hyderabad", "Bangalore", "Warangal", "Hyderabad"
    ],

    "Salary": [
        30000, 35000, np.nan, 40000, 28000,
        45000, 38000, np.nan, 30000, 35000
    ]
}

df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)

print("\n--- DataFrame Information ---")
df.info()

print("\n--- Statistical Description ---")
print(df.describe())

print("\n--- Name and City Columns ---")
print(df[["Name", "City"]])

salary_mean = df["Salary"].mean()

df["Salary"] = df["Salary"].fillna(salary_mean)

print("\n--- After Filling Missing Salary Values ---")
print(df)

df = df.drop_duplicates()

print("\n--- After Removing Duplicate Rows ---")
print(df)

df = df.drop("City", axis=1)

print("\n--- Final DataFrame After Dropping City ---")
print(df)
```
